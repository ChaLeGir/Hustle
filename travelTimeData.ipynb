```python
#Imports the things

import ipywidgets as widgets
from IPython.display import display, clear_output
```

```python
def edit_objects(objects, attrs=None):
    """
    Display an editable table for a list of Python objects.
    Edits are applied back to the original objects when Save is clicked.
    """

    if not objects:
        print("No objects to edit.")
        return

    # Default: use attributes from the first object's __dict__
    if attrs is None:
        attrs = list(objects[0].__dict__.keys())

    grid = widgets.GridBox(
        layout=widgets.Layout(
            grid_template_columns=f"repeat({len(attrs) + 1}, 180px)",
            grid_gap="6px"
        )
    )

    cells = []

    # Header row
    children = [widgets.HTML("<b>Index</b>")]
    children += [widgets.HTML(f"<b>{attr}</b>") for attr in attrs]

    # Data rows
    for i, obj in enumerate(objects):
        children.append(widgets.Label(str(i)))

        row_cells = {}
        for attr in attrs:
            value = getattr(obj, attr, "")

            cell = widgets.Text(value=str(value))
            row_cells[attr] = cell
            children.append(cell)

        cells.append(row_cells)

    grid.children = children

    save_button = widgets.Button(description="Save changes", button_style="success")
    output = widgets.Output()

    def save_changes(_):
        type_preservation_failed = False
        for obj, row_cells in zip(objects, cells):
            for attr, cell in row_cells.items():
                old_value = getattr(obj, attr, None)
                new_text = cell.value

                # Try to preserve the original type
                try:
                    if old_value is not None:
                        if isinstance(old_value, bool):
                            new_value = new_text.lower() in ("true", "1", "yes", "y")
                        else:
                            new_value = type(old_value)(new_text)
                    else:
                        new_value = new_text
                        type_preservation_failed = True
                except:
                    new_value = new_text
                    type_preservation_failed = True

                setattr(obj, attr, new_value)

        with output:
            clear_output()
            print(f"Saved changes with type preservation: {not type_preservation_failed}")

    save_button.on_click(save_changes)

    # display(grid, save_button, output)
    display(grid, save_button, output)
```

```python
class Task:
    def __init__(self, name, duration, priority):
        self.name = name
        self.duration = duration
        self.priority = priority

tasks = [
    Task("Read", 45, 2),
    Task("Code", 90, 1),
    Task("Gym", 60, 3)
]
```

```python
edit_objects(tasks)
```

```python
class Location:
    def __init__(self, index):
        self.name = "Location " + str(index)
        self.latitude = None
        self.longitude = None
        self.motorcycle = False
        self.bike = False
        self.car = False

locations = [Location(i) for i in range(10)]
```

```python
edit_objects(locations)
```

```python
locationsReal = [Location(i) for i in range(10)]

edit_objects(locationsReal)
```

```python
edit_objects(locationsReal)
```
